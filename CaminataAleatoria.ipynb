Sea $\{X_n\}_{n \geq 0}$ una caminata aleatoria en los enteros , la cual comenzareos en 0 es decir:
$$
X_0 = 0
$$
La cual para el tiempo n se encontrara en la posicion:
$$
X_n = X_{n-1} + d_n
$$
Es decir la nueva posicion dependera unicamente de la posicion anterior y que se cumpla el evento de "direccion" $d_n$:
$$
d_n =
\begin{cases}
+1 & \text{con probabilidad p (Derecha) } \\
-1 & \text{con probabilidad q (Izquierda)} 
\end{cases}
$$
Siempre que se cumpla que:
$$
p + q = 1
$$
Es decir  despues de n pasos podremos expresar la posicion como:
$$
X_n = \sum_{i=1}^{n} \xi_i
$$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def CA(n,p):
    q=1-p
    m=np.random.choice([1,-1],size=n,p=[p,q])
    direccion=np.cumsum(m)
    return direccion
```

```python
def GR(n,p):
    d=CA(n,p)
    plt.figure(figsize=(10,6))
    plt.plot(d)
    plt.title(f"Caminata Aleatoria (p={p})")
    plt.xlabel("Tiempo")
    plt.ylabel("Posicion")
    plt.grid()
    plt.show()
    AR(n,p)
```

```python
def GVR(n,p,r):
    plt.figure(figsize=(10,6))
    for i in range(r):
        d=ca(n,p)
        plt.plot(d, alpha=0.7)
    plt.title(f"Varios Resultados (p={p})")
    plt.xlabel("Tiempo")
    plt.ylabel("Posicion")
    plt.grid()
    plt.show()
    AR(n,p)
```

```python
def AR(n, p):
    esperanza=n*(2*p-1)
    varianza=n*(1-(2*p-1)**2)
    print("\n--- ANÁLISIS DEL MODELO ---\n")
    print(f"Numero de pasos: {n}")
    print(f"Probabilidad p: {p}")
    print(f"Esperanza teórica E[X_n] = {esperanza:.4f}")
    print(f"Varianza teórica Var(X_n) = {varianza:.4f}\n")
    if p==0.5:
        print("Caso (p = 0.5):")
        print("En general no existe tendencia clara hacia algun lado en especifico .")
        print("Las trayectorias oscilan alrededor de cero.")
    elif p > 0.5:
        print("Caso (p > 0.5):")
        print("Existe una tendencia promedio hacia arriba en la grafica (Derecha en la linea real o los enteros)")
    else:
        print("Caso (p < 0.5):")
        print("Existe una tendencia promedio hacia abajo en la grafica (Izquierda en la linea real o los enteros)")
```
