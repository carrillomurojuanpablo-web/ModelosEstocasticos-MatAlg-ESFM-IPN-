```python
## funcion para calcular el maximo común divisor (MCD)
def MCD(a, b): 
    while b!=0: #repetimos el algoritmo de la division hasta que el divisor sea 0 
        c=b #guardamos temporalmente el valor del divisor b 
        b=a%b#actualizamos el valor de b con el nel residuo de dividir a entre b 
        a=c#actualizamos a con el valor de c , tal como lo pide el algoritmo de la division
    return a #cuando terminamos de repetir el algoritmo retornamos el MCD

#mediante esta funcion calculamos el valor de del modulo inverso (d) mediante iteraciones
def ModInv(e, phi):
    d=1 #inicializamos d en 1 
    while (e*d)%phi!=1: #buscamos el valor de d tal que se cumpla la congruencia 
        d+=1 #incrementaremsos el valor de d hasta que se cumpla la congruencia
    return d

#esta funcion nos permite exponenciar un numero de forma mas eficiente reduciendo las multiplicaciones
def ExpModRap(base, exp, mod):
    r=1 #iniciamos el resultado en uno (neutro multiplicativo)
    base=base%mod #reducimos la base al modulo
    while exp>0: #itermamos mientras el exponente sea mayor a 0
        if exp% 2==1: #si el exponente es impar 
            r =(r*base) % mod #actualizamos el resultado
        exp=exp>>1  #dividimos el exponente por 2 usando desplazamiento
        base=(base*base) %mod #elevamos la base al cuadrado
    return r

#separaremos el texto letra a letra y las convertiremos a numeros
def TexNum(texto):
    numeros=[-(ord(' ')-ord('A'))if letra==' 'else ord(letra.upper())-ord('A')+1 for letra in texto]
    print("Arreglo de números:", numeros)
    return numeros

#funcion para convertir el arreglo de numeros en una palabra
def NumTex(arreglo):
    texto=''.join([' ' if num < 0 else chr(num + ord('A') - 1) for num in arreglo])
    print("Texto descifrado:", texto)
    return texto

#funcion para cifrar cada elemento del arreglo de numeros
def cifrar(arreglo, e, n):
    cifrado = [num if num < 0 else ExpModRap(num, e, n) for num in arreglo]
    print("Arreglo cifrado:", cifrado)
    return cifrado

#funcion para descifrar los elementos del arreglo 
def descifrar(arreglo, d, n):
    descifrado = [num if num < 0 else ExpModRap(num, d, n) for num in arreglo]
    print("Arreglo descifrado:", descifrado)
    return descifrado

#nos aseguraremos de que  nos proporcionen un numero entero , si me da tiempo compruebo tambien que sea primo
def entero(mensaje):
    while True:
        try:
            return int(input(mensaje))
        except ValueError:
            print("Error: Ingresa un número entero válido.")

#funcion para generar los elementos que nesesitamos (n,phi)
def generar():
    while True:
        p= entero("Bienvenido !!! (ingresa un numero primo p): ")
        q =entero("(ingresa un numero primo entero q): ")
        if p==q: #verificamos que p y q sean distintos , se supone que deberia verificar que ambos sean primos sin embargo eso creo implicaria mas trabajo de procesamiento en el caso que sean numeros realmente grandes
            print("Error: p y q deben ser números primos distintos.")
        else:
            break

    n=p*q   #calculamos los valores de n y phi 
    print("n =",n)
    phi=(p-1) * (q-1)
    print("phi =",phi)

    while True:#solicitamos un entero e que sea cooprimo y este en el intervalo dado
        e=entero(f"(ingresa un numero entero e tal que 1<e<{phi} y sea cooprimo de {phi} : ")
        if 1<e<phi and MCD(phi,e) == 1: 
            d=ModInv(e,phi)
            break
        else:
            print("Error: e no cumple con las condiciones o no es coprimo con phi. Intenta nuevamente.")

    print("Clave pública: (",n,",",e,")")
    print("Clave privada: (",d,",",n,")")
    return p, q, n, phi, e, d

#ciframos un mensaje usando las funciones que escribimos antes 
def CIFRADO():
    p, q, n, phi, e, d = generar()
    texto=input("Ingrese la palabra o frase a cifrar: ")
    arreglo=TexNum(texto)
    cifrado=cifrar(arreglo,e,n)

#desciframos el mensaje con las funciones que hicimos 
def DESCIFRAR():
    while True:
        try:
            d,n=map(int, input("Ingrese la clave privada (d, n): ").split(','))
            arreglo=list(map(int, input("Ingrese el arreglo cifrado (separado por comas): ").split(',')))
            break
        except ValueError:
            print("Error: Formato inválido. Intenta nuevamente.")

    descifrado=descifrar(arreglo,d,n)
    NumTex(descifrado)

def EjecutarRSA():
    while True:
        print("\nMenu de Opciones:")
        print("1.Cifrar mensaje")
        print("2.Descifrar mensaje")
        print("0.Salir")
        try:
            opcion = int(input("Seleccione una opción: "))
        except ValueError:
            print("Opción no válida.")
            continue

        if opcion==1:
            CIFRADO()
        elif opcion==2:
            DESCIFRAR()
        elif opcion==0:
            print("Saliendo...")
            break
        else:
            print("Opción no válida.")

EjecutarRSA() #ejecutamos el programa
```
